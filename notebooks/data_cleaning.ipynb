```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
frequency_mapping = {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
df['previous_purchases_band'] = pd.cut(
    df['previous_purchases'],
    bins=[0, 10, 25, 40, 50],
    labels=['New/Low', 'Emerging', 'Established', 'Heavy'],
    include_lowest=True
)
```

```python
df[['previous_purchases', 'previous_purchases_band']].head(10)
```

```python
df[['previous_purchases', 'previous_purchases_band']].head(10)
```

```python
df['discount_flag'] = (df['discount_applied'] == 'Yes')
df['is_subscriber'] = (df['subscription_status'] == 'Yes')
```

```python
df[['discount_applied', 'discount_flag', 'subscription_status', 'is_subscriber']].head(10)
```

```python
df['shipping_type'].unique()
```

```python
shipping_map = {
    'Express': 'Express',
    'Next Day Air': 'Express',
    'Standard': 'Standard',
    'Free Shipping': 'Standard',
    '2-Day Shipping': 'Standard',
    'Store Pickup': 'Other'
}

df['shipping_speed'] = df['shipping_type'].map(shipping_map).fillna('Other')
```

```python
df[['shipping_type', 'shipping_speed']].value_counts()
```

```python
df['payment_method'].unique()
```

```python
payment_map = {
    'PayPal': 'Digital Wallet',
    'Venmo': 'Digital Wallet',
    'Credit Card': 'Card',
    'Debit Card': 'Card',
    'Cash': 'Cash',
    'Bank Transfer': 'Other'
}

df['payment_group'] = df['payment_method'].map(payment_map).fillna('Other')
```

```python
df[['payment_method', 'payment_group']].value_counts()
```

```python
df['purchase_frequency_days'].value_counts().sort_index()
```

```python
df['frequency_band_days'] = pd.cut(
    df['purchase_frequency_days'],
    bins=[0, 7, 30, 90, 365],
    labels=['Very Frequent', 'Frequent', 'Occasional', 'Rare'],
    include_lowest=True
)
```

```python
df[['purchase_frequency_days', 'frequency_band_days']].value_counts().sort_index()
```

```python
def normalize(s):
    return (s - s.min()) / (s.max() - s.min())

norm_amount = normalize(df['purchase_amount'])
norm_prev = normalize(df['previous_purchases'])
norm_rating = normalize(df['review_rating'])

df['value_score'] = (
    (norm_amount * 0.4) +
    (norm_prev * 0.35) +
    (norm_rating * 0.25)
)

# Discount-reliant customers get a slight penalty (10%) since their spend skews toward markdowns
df.loc[df['discount_flag'], 'value_score'] *= 0.9
```

```python
df['value_score'].describe()
```

```python
df['value_segment'] = pd.qcut(
    df['value_score'],
    q=4,
    labels=['Low', 'Mid', 'High', 'VIP']
)
```

```python
df['value_segment'].value_counts()
```

```python
df.to_csv('customer_shopping_trends_cleaned.csv', index=False)
df.columns
```
